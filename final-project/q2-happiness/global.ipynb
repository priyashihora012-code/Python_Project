```python
plt.figure(figsize=(10, 8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot = True,
    cmap="coolwarm"
)

plt.title("Corelation Heatmap")
plt.show()
```

Observation:

1. GDP per capita has a strong positive correlation (0.797) with Happiness Score.
2. Social support also has a strong positive correlation (0.783).
3. Healthy life expectancy shows a strong positive relationship (0.786).
4. Freedom to make life choices has a moderate positive correlation (0.576).
5. Generosity has a very weak correlation (0.090).
6. Perceptions of corruption has a moderate positive correlation (0.401).

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="GDP per capita",
    y="Score"
)

plt.title("GDP per Capita vs Happiness Score")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Social support",
    y="Score"
)

plt.title("Social Support vs Happiness Score")
plt.xlabel("Social Support")
plt.ylabel("Happiness Score")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Healthy life expectancy",
    y="Score"
)

plt.title("Healthy Life Expectancy vs Happiness Score")
plt.xlabel("Healthy Life Expectancy")
plt.ylabel("Happiness Score")

plt.show()
```

```python
top10 = df.nlargest(10, "Score")

plt.figure(figsize=(10,6))

sns.barplot(
    data=top10,
    x="Score",
    y="Country or region",
    hue="Country or region",
    palette="viridis",
    legend=False
)

plt.title("Top 10 Happiest Countries")
plt.xlabel("Happiness Score")
plt.ylabel("Country or Region")

plt.show()
```

```python
bottom10 = df.nsmallest(10, "Score")

plt.figure(figsize=(10,6))

sns.barplot(
    data=bottom10,
    x="Score",
    y="Country or region",
    hue="Country or region",
    palette="magma",
    legend=False
)

plt.title("Bottom 10 Happiest Countries")
plt.xlabel("Happiness Score")
plt.ylabel("Country or Region")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Score"],
    bins=20,
    kde=True,
    color="skyblue"
)

plt.title("Distribution of Happiness Scores")
plt.xlabel("Happiness Score")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["Score"],
    color="lightgreen"
)

plt.title("Box Plot of Happiness Scores")
plt.xlabel("Happiness Score")

plt.show()
```

```python
top5 = df.nlargest(5, "Score")

plt.figure(figsize=(7,7))

plt.pie(
    top5["Score"],
    labels=top5["Country or region"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Happiest Countries")
plt.show()
```

1. GDP per Capita has a strong positive relationship with Happiness Score.

2. Countries with better Social Support generally have higher Happiness Scores.

3. Healthy Life Expectancy also has a positive impact on Happiness Score.

4. Finland is the happiest country in the dataset, while the bottom-ranked countries have much lower Happiness Scores.

5. Most countries have Happiness Scores between 4.5 and 6.5.

6. The Heatmap shows that GDP per Capita, Social Support, and Healthy Life Expectancy are the strongest factors affecting Happiness.

Conclusion

This project analyzed the World Happiness Report 2019 dataset using Python, Pandas, Matplotlib, and Seaborn.

The analysis shows that GDP per Capita, Social Support, and Healthy Life Expectancy have a strong positive relationship with Happiness Score. Countries with better economic conditions, stronger social support, and healthier lives generally report higher happiness levels.
