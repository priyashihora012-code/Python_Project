```python
df["Sex"]= df["Sex"].str.strip().str.lower()
```

```python
df["Embarked"].unique()
```

```python
df["Embarked"]=df["Embarked"].str.strip().str.upper()
```

```python
df["Embarked"].unique()
```

```python
df["Age"].isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Age"].isnull().sum()
```

```python
df["Fare"].isnull().sum()
```

```python
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
```

```python
df["Fare"].isnull().sum()
```

```python
df["Embarked"].isnull().sum()
```

```python
df["Embarked"].isnull().sum()
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df["Embarked"].isnull().sum()
```

```python
df["Cabin"].isnull().sum()
```

```python
df.drop("Cabin", axis=1, inplace=True)
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Survived")

plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Sex", hue="Survived")

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Pclass", hue="Survived")

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x="Age", bins=20, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x="Fare", bins=20, kde=True)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, y="Fare")

plt.title("Box Plot of Fare")
plt.ylabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(6,6))

df["Survived"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    labels=["Not Survived", "Survived"]
)

plt.title("Survival Percentage")
plt.ylabel("")

plt.show()
```

## Final Observations

* The dataset contained missing values, duplicate records, and inconsistent data, which were cleaned before analysis.
* Female passengers had a higher survival rate than male passengers.
* Passenger class influenced survival, with higher-class passengers generally showing better survival.
* Most passengers belonged to a limited age range, as shown in the age distribution.
* The Fare column contained several high-value outliers.
* The correlation heatmap showed the relationships between the numeric variables in the dataset.
