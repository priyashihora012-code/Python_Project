```python
(df == -200).sum()
```

```python
df.replace(-200, np.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

df[numeric_columns] = df[numeric_columns].fillna(df[numeric_columns].mean())
```

```python
df.isnull().sum()
```

```python
df.drop(columns=["Unnamed: 15", "Unnamed: 16"], inplace=True)
```

```python
df.dropna(subset=["Date", "Time"], inplace=True)
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["CO(GT)"], bins=30, kde=True)

plt.title("Distribution of CO(GT)")
plt.xlabel("CO(GT)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["CO(GT)"])

plt.title("Boxplot of CO(GT)")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x="T", y="CO(GT)", data=df)

plt.title("Temperature vs CO(GT)")
plt.xlabel("Temperature")
plt.ylabel("CO(GT)")

plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
```

```python
plt.figure(figsize=(12,5))

plt.plot(df["Date"], df["CO(GT)"])

plt.title("CO(GT) Levels Over Time")
plt.xlabel("Date")
plt.ylabel("CO(GT)")

plt.xticks(rotation=45)

plt.show()
```

## Final Observations

- The dataset contained missing values, empty columns, and invalid values (-200), which were cleaned before analysis.
- Most CO(GT) values were concentrated in a lower range, while a few high values indicated higher pollution levels.
- The boxplot revealed the presence of outliers in the CO(GT) data.
- The correlation heatmap showed relationships among different air quality and weather variables.
- The scatter plot indicated no strong linear relationship between temperature and CO(GT).
- The line plot showed that CO(GT) levels varied over time, indicating fluctuations in air quality.

## 💡 Key Insight

Despite both being environmental variables, the scatter plot showed **no strong linear relationship** between `Temperature` and `CO(GT)` — meaning pollution spikes were driven more by **traffic and industrial activity** than by weather conditions alone.

The time-series trend also revealed that `CO(GT)` levels fluctuated in **cyclical patterns** rather than showing a one-directional improvement or decline.

> 🔑 **Takeaway:** Air quality here is primarily influenced by **recurring human activity cycles** (like daily traffic peaks) rather than seasonal weather change.
